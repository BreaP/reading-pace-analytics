```python
import random

energy = 80
simoleons = 200
fun = 50
story_log = []

print("--- SIM STATUS ---")
print(f"Simoleons: {simoleons} | Energy: {energy} | Fun: {fun}")
print("------------------\n")


print("""You wake up to your alarm shaking the nightstand. You have a morning shift at the tech support center,
but your social battery is empty. Do you:
1) Go to work anyway
2) Pull a 'sick day' and play video games all morning?""")

choice = input("Enter 1 or 2: ")

if choice == "1":
  simoleons += 100
  energy -= 30
  story_log.append("Morning: You dragged yourself to work and made some bank, but you are exhausted.")
elif choice == "2":
  simoleons -= 50
  fun += 40
  story_log.append("Morning: You called in sick and gamed in your pajamas all morning. Worth it.")
else:
  print("Invalid choice! You panicked, froze up, and did nothing all morning.")
  story_log.append("You paralyzed yourself with indecision and wasted the morning.")


afternoon_events = ["Afternoon: A fire started in the kitchen!", "Afternoon: You found 20 simoleons on the sidewalks!", "Afternoon: Your neighbor knocked on the door to gossip."]

event_happened = random.choice(afternoon_events)

print(event_happened)
story_log.append(event_happened)

print("\n===============================")
print("--- FINAL END OF DAY REPORT ---")
print("===============================")

print(f"Final Simoleons: {simoleons}")
print(f"Final Energy: {energy}")
print(f"Final Fun: {fun}\n")

print("--- Day Timeline Summary ---")
for item in story_log:
  print(f"- {item}")
print("=================================")
```
