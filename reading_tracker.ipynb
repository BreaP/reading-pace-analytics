```python
name = "Breana"

total_goal = 100
weeks_remaining = 31
historical_books_count = 14

reading_history = []

def add_book(title, author, genre, pages, rating):
  new_book = {"title": title,
               "author": author,
               "genre": genre,
               "pages": pages,
               "rating": rating}
  reading_history.append(new_book)

add_book("Heart Of A Goon", "Jahquel J", "Urban Fiction", 455, 5)
add_book("Never faking about you", "Aubry J", "Romance", 330, 5)

books_completed = len(reading_history) + historical_books_count
books_remaining = total_goal - books_completed
books_per_week = books_remaining / weeks_remaining

expected_books_by_now = total_goal * ((52 - weeks_remaining) / 52)
left_to_read = expected_books_by_now - books_completed


print(f"{name}, you have {books_per_week:.2f} books to read per week to accomplish your goal.")
print(f"Total books logged in history: {len(reading_history)}")


if books_completed >= expected_books_by_now:
  print(f"{name} you are ahead of schedule!")
else:
  print(f"{name}, you have {left_to_read:.1f} more books to reach your goal! Kick it up a notch.")
```
